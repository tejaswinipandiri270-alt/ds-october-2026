```python
missing_pct = (df.isnull().mean() * 100).round(1)
missing_pct[missing_pct > 0].sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

```

```python
missing_pct[missing_pct > 0].sort_values().plot(kind="barh", title="% missing per column")
plt.xlabel("% missing")
plt.show()
```
